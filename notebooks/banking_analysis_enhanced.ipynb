## Setup and Imports

```python
# Import required libraries
import sys
sys.path.append('../src')

from bank_analysis import BankDataFetcher, BankAnalyzer, BankVisualizer
from bank_analysis.utils import load_config, get_date_range_description, format_percentage

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Configure pandas display options
pd.set_option('display.max_columns', None)
pd.set_option('display.precision', 4)

print("✓ Libraries imported successfully")
```

## Load Configuration

```python
# Load configuration from config.yaml
config = load_config('../config.yaml')

tickers = config['tickers']
start_date = config['date_range']['start']

print(f"Analyzing tickers: {', '.join(tickers)}")
print(f"Date range: {start_date} to today")
```

## 1. Data Fetching

Fetch historical stock data for major banking institutions using Yahoo Finance API.

```python
# Initialize data fetcher
fetcher = BankDataFetcher(tickers=tickers, start_date=start_date)

# Fetch data
data = fetcher.fetch_data()

print(f"\nData shape: {data.shape}")
print(f"Date range: {data.index[0]} to {data.index[-1]}")
print(f"\nFirst few rows:")
data.head()
```

```python
# Display basic information about the data
print("Data Info:")
print(f"Total trading days: {len(data)}")
print(f"Missing values per ticker:")
print(data.isnull().sum())
print(f"\nData types:")
print(data.dtypes)
```

## 2. Price Analysis

Visualize price trends over time.

```python
# Initialize visualizer
visualizer = BankVisualizer(data)

# Plot price trends
visualizer.plot_prices(
    title="Banking Stocks: Adjusted Close Price Trends",
    save_path='../output/plots/price_trends.png'
)
```

## 3. Returns Analysis

Calculate and analyze daily returns.

```python
# Initialize analyzer
analyzer = BankAnalyzer(data)

# Calculate returns
returns = analyzer.calculate_returns(method='pct_change')

print(f"Returns shape: {returns.shape}")
print(f"\nFirst few returns:")
returns.head()
```

```python
# Summary statistics for returns
summary_stats = analyzer.get_summary_statistics()
print("Returns Summary Statistics:")
summary_stats
```

```python
# Update visualizer with returns data
visualizer.returns = returns

# Plot returns distribution
visualizer.plot_returns_distribution(
    save_path='../output/plots/returns_distribution.png'
)
```

## 4. Correlation Analysis

Analyze correlations between different banking stocks.

```python
# Calculate correlation matrix
correlation_matrix = analyzer.get_correlation_matrix()

print("Correlation Matrix of Daily Returns:")
correlation_matrix
```

```python
# Visualize correlation heatmap
visualizer.plot_correlation_heatmap(
    correlation_matrix,
    title="Banking Stocks: Correlation Matrix of Daily Returns",
    save_path='../output/plots/correlation_heatmap.png'
)
```

## 5. Cumulative Returns

Calculate and visualize cumulative returns to see total performance over time.

```python
# Calculate cumulative returns
cumulative_returns = analyzer.calculate_cumulative_returns()

print("Cumulative Returns (Final Values):")
print((cumulative_returns.iloc[-1] * 100).round(2))
print("\nNote: Values shown as percentage")
```

```python
# Plot cumulative returns
visualizer.plot_cumulative_returns(
    cumulative_returns,
    title="Banking Stocks: Cumulative Returns Over Time",
    save_path='../output/plots/cumulative_returns.png'
)
```

```python
# Find best and worst performers
best, worst = analyzer.find_best_worst_performers()
best_return = cumulative_returns.iloc[-1][best]
worst_return = cumulative_returns.iloc[-1][worst]

print(f"🏆 Best Performer: {best} with {format_percentage(best_return)} cumulative return")
print(f"📉 Worst Performer: {worst} with {format_percentage(worst_return)} cumulative return")
```

## 6. Volatility Analysis

Analyze the volatility (risk) of each banking stock.

```python
# Calculate rolling volatility
volatility = analyzer.calculate_volatility(window=30)

print("Current Volatility (30-day rolling):")
print(volatility.iloc[-1].round(6))
```

```python
# Plot volatility over time
visualizer.plot_volatility(
    volatility,
    title="Banking Stocks: 30-Day Rolling Volatility",
    save_path='../output/plots/volatility.png'
)
```

## 7. Risk-Adjusted Performance

Calculate Sharpe ratios to evaluate risk-adjusted returns.

```python
# Calculate Sharpe ratios
sharpe_ratios = analyzer.calculate_sharpe_ratio(risk_free_rate=0.02)

print("Sharpe Ratios (Risk-Adjusted Performance):")
print(sharpe_ratios.round(3))
print("\nHigher Sharpe ratio indicates better risk-adjusted performance")
print(f"Best Sharpe Ratio: {sharpe_ratios.idxmax()} ({sharpe_ratios.max():.3f})")
```

## 8. Price Statistics

Summary of key price metrics.

```python
# Get price statistics
price_stats = analyzer.get_price_statistics()

stats_df = pd.DataFrame(price_stats).round(2)
print("Price Statistics Summary:")
stats_df
```

## 9. Comprehensive Dashboard

Create a single comprehensive dashboard with all key visualizations.

```python
# Create comprehensive dashboard
visualizer.create_dashboard(
    analyzer,
    save_path='../output/plots/comprehensive_dashboard.png'
)
```

## 10. Save Results

Export processed data and results for further analysis.

```python
# Save processed data
data.to_csv('../data/processed/bank_prices.csv')
returns.to_csv('../data/processed/bank_returns.csv')
cumulative_returns.to_csv('../data/processed/bank_cumulative_returns.csv')
correlation_matrix.to_csv('../data/processed/bank_correlation_matrix.csv')

print("✓ All data saved successfully to data/processed/")
```
